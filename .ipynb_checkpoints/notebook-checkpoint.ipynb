### 1. On the Meaning of $\text{OLS}$ Parametric Regression

</br>

$$\hat{y}_j = \beta_0 + \sum_{i=1}^{p} \beta_i \, x_{ij} \quad;\qquad\forall\;j$$

Where:
$$
\begin{aligned}
&\hat{y} &\equiv& &\text{Predicted Outcome}& \qquad\qquad &\beta &\equiv& &\text{Trainable Parameter}& \\\\
&j &\equiv& &\text{Datapoint Index}& \qquad\qquad &n &\equiv& &\text{Sample Size (i.e. Datapoint Count)}& \\\\
&i &\equiv& &\text{Predictor Index}& \qquad\qquad &p &\equiv& &\text{Predictor Count}&
\end{aligned}
$$

</br>
An Ordinary Least Squares ($\text{OLS}$) solution dictates that:

$$
\begin{align}
    J\,(\beta_0, \,\beta_1, \,\dots\,, \,\beta_p) = \sum_{j=1}^{n} e_j^2
\end{align}
$$

</br>

Where:

$$
\begin{align}
    e_j^2 &= (y_j - \hat{y}_j)^2 \\\\
    e_j^2 &= \left[y_j - \beta_0 - \sum_{i=1}^{p} \beta_i \, x_{ij}\right]^2 \\\\
\end{align}
$$

</br>

*Since $J\,(\cdot)$ is made up of affines (i.e. linear expressions of $\beta_i$) which are convex functions themselves, their squares are also convex that open upward due to having a positive $\beta_i^2$ terms. Making $J_{min}\,(\cdot)$ occur at a grad of zero:*

$$\frac{\partial\,J\,(\cdot)}{\partial\,\beta_i} = 0 \qquad\longrightarrow\qquad 2\,\sum_{j=1}^n \, e_j \, \frac{\partial\,e_j}{\partial\,\beta_i} = 0 \qquad\longrightarrow\qquad \sum_{j=1}^n \, e_j \, \frac{\partial\,e_j}{\partial\,\beta_i} = 0  \quad;\qquad\forall\;i$$

</br>

$$
\begin{align}
    \sum_{j=1}^n \, x_{ij} \, \left(y_j - \beta_0 - \sum_{i=1}^{p} \beta_i \, x_{ij}\right) &= 0\quad;\qquad&\forall\;&i\neq0 \\\\
    \sum_{j=1}^n \, \left(y_j - \beta_0 - \sum_{i=1}^{p} \beta_i \, x_{ij}\right) &= 0\quad;\qquad\;&&i=0
\end{align}
$$

</br>

Therefore, we have two conditions for an $\text{OLS}$ solution:

$$
\begin{align}
    \sum_{j=1}^n \, x_{ij} \, e_{j} = 0 \qquad\qquad \sum_{j=1}^n \, e_{j} = 0
\end{align}
$$

</br>

Interestingly, this can be translated to:

$$
\begin{align}
    \left<x_{i},\, e\right> = 0 \qquad\qquad \left<1,\, e\right> = 0
\end{align}
$$

</br>

*Given that our dataset is represented by $n$ datapoints in a $p$-dimensional feature-space, it can be considered as a matrix that is transforming such space so that its datapoints get projected onto an alternate space of $n$ dimensions, and that the original features of the dataset are now our transformation basis vectors between the two spaces.*

*Therefore, and due to the derived $\text{OLS}$ conditions, a parametric linear regression model finds its $n$-dimensional $\text{OLS}$ error vector $e$ that is normal to the sub-plane formed by a n-dimensional vector of ones (i.e. indicating the model's bias term), in addition to all the dataset's transformational basis' vectors (i.e. feature dimensions in the original space).*

*Thus, an $\text{OLS}$ solution in the above context boils down to finding a normal vector to a specific sub-space which does not require iterative parameter optimization (i.e. gradient descent).*

### 2. On the Meaning of $\text{MLE}$ Parametric Regression

</br>

$$
\begin{aligned}
y_j &= \hat{y}_j + e_j \quad&;\qquad\forall\;j
\end{aligned}
$$

Where:
$$
\begin{aligned}
&{y} &\equiv& &\text{Actual Outcome}& \qquad\qquad &e_j &\equiv& &\text{Prediction Error}&
\end{aligned}
$$

</br>

Substituting with the parametric evaluation of $\hat{y}_j$:

$$
\begin{aligned}
y_j &= \left(\beta_0 + \sum_{i=1}^{p} \beta_i \, x_{ij}\right) + e_j \quad&;\qquad\forall\;j
\end{aligned}
$$

</br>

Since prediction errors are already irreducible by the regression model producing them, they can only be indepdently distributed in a random sequence. So, with the predicted values ~$\beta_0 + \sum_{i=1}^{p} \beta_i \, x_{ij}$~ being deterministic and the prediction errors $e_j$ identically following a certain probability density expression, the actual outcome value $y_j$ must follow that same desnity in a conditional construct (i.e. likelihood).

For the sake of argument, let's <span style="color: red;">***assume***</span> our prediction error term $e_j$ follows a Gaussian distribution of a zero mean and a variance $\sigma^2$:

$$
\begin{aligned}
e_j \sim \mathcal{n}(0, \sigma^2) \quad&;\qquad\forall\;j
\end{aligned}
$$

</br>

Therefore, the actual outcomes $y_j$ must follow a conditional density of:

$$
\begin{aligned}
p\,(y_j\,|\,x_j;\,\beta_0,\,\beta_1,\,\dots,\beta_p,\,\sigma^2) = \frac{1}{\sqrt{2\,\pi\,\sigma^2}}\,\exp\,\left[-\frac{\left(y_j - \beta_0 - \sum_{i=1}^{p} \beta_i \, x_{ij}\right)^2}{2\,\sigma^2}\right]
\end{aligned}
$$

</br>

Thus, the likelihood $\mathcal{L}$ of our trainable parameters $\beta_i$ explaining the normal distribution of actual outcomes $y_j$ about the model's predicted $\hat{y}_j$ at the variance of $\sigma^2$ is:

$$
\begin{aligned}
\mathcal{L}\,(\beta_0,\,\beta_1,\,\dots,\beta_p,\,\sigma^2) &= \prod_{j=1}^{n}\,p\,(y_j\,|\,x_j;\,\underline{\beta},\,\sigma^2) \\\\
\mathcal{L}\,(\beta_0,\,\beta_1,\,\dots,\beta_p,\,\sigma^2) &= \left(\frac{1}{2\,\pi\,\sigma^2}\right)^{\frac{n}{2}}\,\exp\,\left[-\frac{1}{2\,\sigma^2}\,\sum_{j=1}^{n}\,\left(y_j - \beta_0 - \sum_{i=1}^{p} \beta_i \, x_{ij}\right)^2\right]
\end{aligned}
$$

</br>

Simplifying the expression while preserving its characteristics by taking the natural logarithm, we obtain the log-likelihood $\ell$:

$$
\begin{aligned}
\ell\,(\beta_0,\,\beta_1,\,\dots,\beta_p,\,\sigma^2) &= \ln\,\mathcal{L}\,(\beta_0,\,\beta_1,\,\dots,\beta_p,\,\sigma^2) \\\\
\ell\,(\beta_0,\,\beta_1,\,\dots,\beta_p,\,\sigma^2) &= -\frac{n}{2}\ln\left(2\,\pi\,\sigma^2\right)\,-\,\frac{1}{2\,\sigma^2}\,\sum_{j=1}^{n}\,\left(y_j - \beta_0 - \sum_{i=1}^{p} \beta_i \, x_{ij}\right)^2
\end{aligned}
$$

</br>

Therefore, the regression problem as a Maximum Likelihood Estimation ($\text{MLE}$) one becomes a question of optimizing our trainable parameters $\beta_i$ so that the actual outcomes $y_j$ are best explained by the Gaussian distribution of variance $\sigma^2$ about the model's predicted $\hat{y}_j$.

Since the second part of $\ell$ is just a negative summation of squared affines, it can be deduced that the log-likelihood is convex curving downward with a maximum value. Therefore, a zero grad with respect to $\beta_i$ yields our conditions for maximum likelihood and subsequently, our optimized parameters $\beta_i$ that best represent the actual data $y_j$ given our error $e_j$ Gaussian distribution assumption.

$$\frac{\partial\,\ell\,(\cdot)}{\partial\,\beta_i} = 0$$

</br>

This gives us:

$$
\begin{align}
    \sum_{j=1}^n \, x_{ij} \, \left(y_j - \beta_0 - \sum_{i=1}^{p} \beta_i \, x_{ij}\right) &= 0\quad;\qquad&\forall\;&i\neq0 \\\\
    \sum_{j=1}^n \, \left(y_j - \beta_0 - \sum_{i=1}^{p} \beta_i \, x_{ij}\right) &= 0\quad;\qquad\;&&i=0
\end{align}
$$

</br>

Which again can be simplified to:

$$
\begin{align}
    \sum_{j=1}^n \, x_{ij} \, e_{j} = 0 \qquad\qquad \sum_{j=1}^n \, e_{j} = 0
\end{align}
$$

</br>

Giving us the $\text{OLS}$ conditions for the optimal error vector obtained previously:

$$
\begin{align}
    \left<x_{i},\, e\right> = 0 \qquad\qquad \left<1,\, e\right> = 0
\end{align}
$$

</br>

*From this we can deduce that when an $\text{OLS}$ trained parametric regression model has its prediction errors $e_j$ satisfying the **Condition of Normality**, the actual outcomes $y_j$ are expected to be dispersed around $\hat{y}_j$ within an Gaussian envelope of variance $\sigma^2$.*

### 3. On the Meaning of $\text{RMSE}$

</br>

$$
\begin{aligned}
\text{RMSE}\,(y, \hat{y}) &= \sqrt{\frac{1}{n} \, \sum_{j=1}^{n} (y_j - \hat{y}_j)^2}
\end{aligned}
$$

</br>

Conveniently:

$$
\begin{aligned}
\sigma\,(y) &= \sqrt{\frac{1}{n} \, \sum_{j=1}^{n} (y_j - \mu)^2}\\\\
\sigma\,(y) &= \text{RMSE}\,(y, \mu)
\end{aligned}
$$

</br>

Where:
$$
\begin{aligned}
&\mu &\equiv& &\text{Population Mean}& \\\\
&\sigma &\equiv& &\text{Standard Deviation (about $\mu$)}& \\\\
&\text{RMSE} &\equiv& &\text{Effective Error (about $\hat{y}$)}&
\end{aligned}
$$

</br>

*Therefore, if a regression model **happens** to satisfy the Condition of Normality, then, like the standard deviation providing a Quantile Interval* ($\text{QI}$) *to a 1D normal distribution, within which $95\%$ of the values are expected to be dispersed within $\mu\,\pm\,1.96\times\sigma$, the $\text{RMSE}$ can now be interpreted to define a Prediction Interval* ($\text{PI}$) *envelope around the predicted values within which the actual values have a $95\%$ probability of existing:*

$$\text{PI}\,(0.95) \;\approx\; \hat{y} \pm 1.96 \times \text{RMSE}\,(y, \hat{y})$$

</br>

*Thus, the narrower the $\text{PI}$, the more precise our regression model becomes.*

***Note:** A $95\%~\text{PI}$ requires a scaling factor of $1.96$ strictly in the case of the normality condition being satisfied by the regression model. With other prediction error distributions, the $\text{PI}$ bounds about $\hat{y}$ are expected to be formulated differently.*

### 4. On the Meaning of $\text{Var}$

</br>

$$\text{Var}\,(x_1, x_2) = \frac{1}{N} \sum_{j=1}^{N} (x_{1j} - \mu_1)\,(x_{2j} - \mu_2)$$

</br>

Where:

$$
\begin{aligned}
&\text{Var}\,(x_1, x_2) &\equiv& &\text{Covariance between Dimensions}~x_1~\text{and}~x_2& \\\\
&N &\equiv& &\text{Population Count}&
\end{aligned}
$$

</br>

*From the above expression, we can see the Covariance $\text{Var}\,(x_1, x_2)$, as a dispersion measure of datapoint projections across the 2D plane defined by dimensions $x_1$ and $x_2$, can be defined as the average area of rectangles bounded by their respective datapoints and the population's centroid. Notice that an area is a vector quantity, so $\text{Var}\,(x_1, x_2)$ can evaluate to a negative value if the datapoint projections on the $x_1, x_2$ plane have their components, on average, changing their values in opposite directions.*

### 5. On the Meaning of $r$

</br>

$$r\,(x_1, x_2) = \frac{1}{N} \sum_{j=1}^{N} \frac{(x_{1j} - \mu_1)\,(x_{2j} - \mu_2)}{\sigma_1 \, \sigma_2}$$

</br>

Where:

$$
\begin{aligned}
&r\,(x_1, x_2) &\equiv& &\text{Pearson's Correlation between Dimensions}~x_1~\text{and}~x_2&
\end{aligned}
$$

</br>

*Therefore, the Correlation $r\,(x_1, x_2)$, as a normalized dispersion measure of datapoint projections across the 2D plane defined by dimensions $x_1$ and $x_2$, can be used as a unit-less measure to compare the strengths of variational relationships between multiple pairs of dimensions that have different domain understandings (or units of measurement).*

Now, let's derive $r$'s range so we can understand more about its physical nature.

</br>

$$
\begin{align}
r\,(x_1, x_2) &= \frac{1}{N} \sum_{j=1}^{N} \frac{(x_{1j} - \mu_1)\,(x_{2j} - \mu_2)}{\sigma_1 \, \sigma_2} \\\\
r\,(x_1, x_2) &= \frac{\frac{1}{N}\sum_{j=1}^{N}\left[(x_{1j} - \mu_1)\,(x_{2j} - \mu_2)\right]}{\sqrt{\frac{1}{N}\sum_{j=1}^{N}(x_{1j} - \mu_1)^2} \; \sqrt{\frac{1}{N}\sum_{j=1}^{N}(x_{2j} - \mu_2)^2}} \\\\
r\,(x_1, x_2) &= \frac{\sum_{j=1}^{N}\left[(x_{1j} - \mu_1)\,(x_{2j} - \mu_2)\right]}{\sqrt{\sum_{j=1}^{N}(x_{1j} - \mu_1)^2} \; \sqrt{\sum_{j=1}^{N}(x_{2j} - \mu_2)^2}} \\\\
r\,(x_1, x_2) &= \frac{\left<x_{1} - \mu_1,\,x_{2} - \mu_2\right>}{\left\|x_{1} - \mu_1\right\| \; \left\|x_{1} - \mu_1\right\|} = \cos\theta
\end{align}
$$

</br>

Therefore:

$$r(x_1, x_2) \in [-1, 1]$$

</br>

*This means that the correlation $r(x_1, x_2)$ becomes physically the cosine similarity between the population dataset's transformational basis vectors $x_1$ and $x_2$ in a version of the population dataset that is normalized w.r.t its centroid.*

</br>

***Fun Fact:*** 

*In Digital Signal Processing (DSP), the correlation is a measure of similarity between two 1D signals (i.e. sequences), it is computed as their dot-product, which makes it the statistical covariance between pairs of the two 1D populations (i.e. they do not have to be sequences in that context); that is if the signals had their DC components removed first (centroid-normalized).*

*So, instead of being closer to statistical correlation, the DSP correlation measure is actually closer to measuring statistical covariance; and this is not due to the two signals having the same unit of measurement (i.e. Volts), because the natures of both signals could still be different from a domain understanding perspective, but rather due to a DSP application-scope that requires computing the similarity within **only one pair** of 1D signals, which is unlike the statistical application-scope of comparing the relationships between multiple pairs of features.*

### 6. On the Meaning of $\Sigma$ and the $\text{QI}$'s Ellipse

</br>

A Covariance Matrix $\Sigma$ is the symmetric, positive semi-definite one that shows how the datapoints are distributed in the $p$-dimensional space that the dataset exists in. In other words, it makes use of covariance $\text{Var}$ and correlation $r$ measurements that describe the relationships of dimensional feature pairs so that we can describe how our dataset blob is manifested in the feature-space.

For a start, let's describe how a 1D variable $x$ follows a Gaussian distrubtion of mean $\mu$ and variance $\sigma^2$:

$$p(x) = \frac{1}{\sqrt{2\,\pi\,\sigma^2}}\,\exp\left[-\frac{1}{2}\,\frac{(x - \mu)^2}{\sigma^2}\right]$$

</br>

This equation shows us that determining the limits of $x$ for integrating $p(x)$ into a specific percentage is all it takes to determine a specific confidence interval; and since $x$ exists only in the exponent as $\frac{x - \mu}{\sigma^2}$, let's simplify $p(x)$ into:

$$p(c) = \frac{1}{\sqrt{2\,\pi\,\sigma^2}}\,\exp\left[-\frac{1}{2}\,c^2\right]$$

</br>

Where $c$ in that regard, can be considered a positive value used to determine the limits of the $\text{QI}$, given a specific percentage the expression $p(c)$ integrates to. So, let's entertain $c$'s squared expression:

$$
\begin{aligned}
c^2 &= \frac{(x - \mu)^2}{\sigma^2} \\\\
c\,\sigma &= |x - \mu| \\\\
\end{aligned}
$$

</br>

From here, we can see that $c$ can be considered a scaling factor for our $\sigma$ to represent the absolute deviation of $x$ from the population mean $\mu$. To obtain a 1D $\text{QI}$ of $95\%$, $c$ must be set to $1.96$.

Let's now move our discussion to the $p$-dimensional domain to investigate the meaning of our covariance matrix $\Sigma$. As such, terms utilizing the scalar form of $\sigma$ (i.e. $\frac{1}{\sqrt{2\,\pi}\,\sigma}$) can have it substituted with $\sqrt{\det(\Sigma)}$; this is fundamentally because as the $\sigma$ can be considered an unscaled version of the 1D volume a dataset column occupies, the term $\sqrt{\det(\Sigma)}$ actually computes an unscaled $p$-dimensional volume that our datapoints occupy in the dataset's feature-space. Additionally, terms utilizing the inter-dimensional vectorized-form of $\sigma^2$ (i.e. $\frac{(x - \mu)^2}{\sigma^2}$) can now have it substituted with $\Sigma^{-1}$ (i.e. becoming $(\underline{x} - \underline{\mu})^\top\,\Sigma^{-1}\,(\underline{x} - \underline{\mu})$ so it evaluates to a scalar quantity for $p(\underline{x})$). These substitutions give us:

$$p(\underline{x}) = \frac{1}{\sqrt{2\,\pi\,\det(\Sigma)}}\,\exp\left[-\frac{1}{2}\,(\underline{x} - \underline{\mu})^\top\,\Sigma^{-1}\,(\underline{x} - \underline{\mu})\right]$$

</br>

With the current equation the exponent's term $(\underline{x} - \underline{\mu})^\top\,\Sigma^{-1}\,(\underline{x} - \underline{\mu})$ is still the key to determining the $\text{QI}$'s limits when integrating $p(\underline{x})$ to obtain for a certain confidence percentage. Therefore similar to the 1D case: 

$$
c^2 = (\underline{x} - \underline{\mu})^\top\,\Sigma^{-1}\,(\underline{x} - \underline{\mu})
$$

</br>

Let's assume a 2D feature-space to simplify our analysis:

$$
\begin{align}
\Sigma = 
\begin{bmatrix}
    \text{Var}(x_1,\,x_1) & \text{Var}(x_1,\,x_2) \\
    \text{Var}(x_2,\,x_1) & \text{Var}(x_2,\,x_2)
\end{bmatrix} \qquad\longrightarrow\qquad \Sigma = 
\begin{bmatrix}
    \sigma_1^2 & \sigma_1\,\sigma_2\,r(x_1,\,x_2) \\
    \sigma_1\,\sigma_2\,r(x_2,\,x_1) & \sigma_2^2
\end{bmatrix} \qquad\longrightarrow\qquad \Sigma = 
\begin{bmatrix}
    \sigma_1^2 & r\,\sigma_1\,\sigma_2 \\
    r\,\sigma_1\,\sigma_2 & \sigma_2^2
\end{bmatrix}
\end{align}
$$

</br>

Thus:

$$
\begin{align}
\Sigma^{-1} &= \frac{1}{1-r^2}
\begin{bmatrix}
    \frac{1}{\sigma_1^2} & -\frac{r}{\sigma_1\,\sigma_2} \\
    -\frac{r}{\sigma_1\,\sigma_2} & \frac{1}{\sigma_2^2}
\end{bmatrix}
\end{align}
$$

</br>

Therefore:

$$
\begin{align}
c^2 &= \frac{1}{1-r^2} \,
\begin{bmatrix}
    x_1 - \mu_1 &x_2 - \mu_2
\end{bmatrix} \,
\begin{bmatrix}
    \frac{1}{\sigma_1^2} & -\frac{r}{\sigma_1\,\sigma_2} \\
    -\frac{r}{\sigma_1\,\sigma_2} & \frac{1}{\sigma_2^2}
\end{bmatrix} \,
\begin{bmatrix}
    x_1 - \mu_1 \\
    x_2 - \mu_2
\end{bmatrix} \\\\
c^2 &= \frac{1}{1 - r^2}\left[ \left(\frac{x_1 - \mu_1}{\sigma_1}\right)^2 - 2r \left(\frac{x_1 - \mu_1}{\sigma_1}\right)\left(\frac{x_2 - \mu_2}{\sigma_2}\right) + \left(\frac{x_2 - \mu_2}{\sigma_2}\right)^2 \right]
\end{align}
$$

</br>

*The above equation actually represents our ellipse in the 2D feature-space. Setting the scale $c$ to $2.448$ scales our ellipse to bound the $95\%~\text{QI}$ within which our 2D dataset points are dispersed. As we can expect now, a $p$-dimensional dataset will have a $p$-dimensional ellipsoid bounding its $\text{QI}$.*

*But how do we determine the scale $c$ corresponding to a specific $\text{QI}$ percentage? We might do this via trial-and-error over an infinite amount of integrations, which, of course, would take forever, or we can use the critical-value tables of $p$-dimensional chi-squared distributions $\chi_p^2$; these tables are a standard for these types of data analysis tasks. God willing, examining the $\chi_p^2$ mathematical nature is a topic for a later time.*

### 7. On the Meaning of a Parametric Regression Model's $r^2$ Score

</br>

$$
\begin{aligned}
r^2\,(y, \hat{y}) &= \left[\frac{1}{N}\,\sum_{j=1}^{n}\frac{(y_j - \bar{y})\,(\hat{y}_j - \bar{{\hat{y}}})}{\sigma_y\,\sigma_{\hat{y}}}\right]^2 \\\\
r^2\,(y, \hat{y}) &= \left[\frac{\cancel{\frac{1}{n}}\,\sum_{j=1}^{n}\,(y_j - \bar{y})\,(\hat{y}_j - \bar{{\hat{y}}})}{\sqrt{\cancel{\frac{1}{n}}\,\sum_{j=1}^{n}\,(y - \bar{y})}\,\sqrt{\cancel{\frac{1}{n}}\,\sum_{j=1}^{n}\,(\hat{y} - \bar{{\hat{y}}})}}\right]^2 \\\\
r^2\,(y, \hat{y}) &= \frac{\left[\sum_{j=1}^{n}\,(y_j - \bar{y})\,(\hat{y}_j - \bar{{\hat{y}}})\right]^2}{\sum_{j=1}^{n}\,(y - \bar{y})\,\sum_{j=1}^{n}\,(\hat{y} - \bar{{\hat{y}}})}
\end{aligned}
$$

</br>

Where:

$$
\begin{aligned}
\bar{y} &= \frac{1}{n} \, \sum_{j=1}^{n} \, y_j \qquad\equiv \text{Sample Mean of actual outcomes} \\\\
\bar{\hat{y}} &= \frac{1}{n} \, \sum_{j=1}^{n} \, \hat{y}_j \qquad\equiv \text{Sample Mean of predicted outcomes}
\end{aligned}
$$

</br>

Expanding $\hat{y}_j$ and $y_j$, and recalling our optimal solution's $\text{OLS}$ conditions:

$$
\begin{aligned}
&\hat{y}_j &= &\beta_0 + \sum_{i=1}^{p}\,\beta_i\,x_{ij}& \qquad\qquad\qquad &\sum_{j=1}^{n}&\,e_j &= 0& \\\\
&{y}_j = \hat{y}_j + e_j &= &\beta_0 + \sum_{i=1}^{p}\,\beta_i\,x_{ij} + e_j& \qquad\qquad\qquad &\sum_{j=1}^{n}&\,x_{ij}\,e_j &= 0& \quad;\qquad\forall\,i
\end{aligned}
$$

</br>

Now, let's go for obtaining $\bar{y}$ and $\bar{{\hat{y}}}$:

$$
\begin{aligned}
\bar{y} &= \frac{1}{n} \sum_{j=1}^{n}\,\left[\beta_0 + \sum_{i=1}^{p}\,\beta_i\,x_{ij} + e_j\right] \\\\ \bar{y} &= \beta_0 + \sum_{i=1}^{p}\,\left(\beta_i\,\times\,\cancel{\frac{1}{n}\,\sum_{j=1}^{n}\,x_{ij}}^{\bar{x}_i}\right) + \cancel{\frac{1}{n}\,\sum_{j=1}^{n}\,e_j}^0 = \bar{{\hat{y}}} \\\\
\bar{y} &= \bar{{\hat{y}}} = \beta_0 + \sum_{i=1}^{p}\,\beta_i\,\bar{x}_i
\end{aligned}
$$

</br>

Where:

$$
\begin{aligned}
\bar{x} &= \frac{1}{n} \, \sum_{i=1}^{p} \, x_i \qquad\equiv \text{Sample Mean of feature $i$}
\end{aligned}
$$

</br>

Therefore:

$$
\begin{aligned}
r^2\,(y, \hat{y}) &= \frac{\left[\sum_{j=1}^{n}\,(\hat{y}_j - \bar{y})\,\left((\hat{y}_{j} - \bar{y}) + e_j\right)\right]^2}{\sum_{j=1}^{n}\,(y - \bar{y})\,\sum_{j=1}^{n}\,(\hat{y} - \bar{{\hat{y}}})} \\\\
r^2\,(y, \hat{y}) &= \frac{\left[\sum_{j=1}^{n}\,(\hat{y}_j - \bar{y})^2 + \sum_{j=1}^{n}\,e_j\,(\hat{y}_{j} - \bar{y})\right]^2}{\sum_{j=1}^{n}\,(y - \bar{y})\,\sum_{j=1}^{n}\,(\hat{y} - \bar{{\hat{y}}})} \\\\
r^2\,(y, \hat{y}) &= \frac{\left[\text{ESS} + \sum_{j=1}^{n}\,e_j\,(\hat{y}_{j} - \bar{y})\right]^2}{\text{TSS}\,\times\,\text{ESS}} \\\\
\end{aligned}
$$

</br>

Where:

$$
\text{TSS} = \sum_{j=1}^{n}\,(\hat{y}_j - \bar{y})^2 \qquad\qquad \text{ESS} = \sum_{j=1}^{n}\,(\hat{y}_j - \bar{y})^2\\\\
$$

</br>

And for completeness:

$$
\begin{aligned}
\text{RSS} &= \sum_{j=1}^{n}\,e_j^2 \\\\
 &= \sum_{j=1}^{n}\,(y_j - \hat{y}_j)^2\\\\
 &= \sum_{j=1}^{n}\,\left[(y_j - \bar{y}) - (\hat{y}_j - \bar{y})\right]^2\\\\
 &= \sum_{j=1}^{n}\,\left[(y_j - \bar{y})^2 - 2\,(y_j - \bar{y})\,(\hat{y}_j - \bar{y}) + (\hat{y}_j - \bar{y})^2\right]\\\\
 &= \sum_{j=1}^{n}\,\left[(y_j - \bar{y})^2 - 2\,(\hat{y}_j - \bar{y})\,\left((\hat{y}_{j} - \bar{y}) + e_j\right) + (\hat{y}_j - \bar{y})^2\right]\\\\
 &= \sum_{j=1}^{n}\,\left[(y_j - \bar{y})^2 - \cancel{2}\,(\hat{y}_j - \bar{y})^2 + 2\,e_j\,(\hat{y}_j - \bar{y})^2 + \cancel{(\hat{y}_j - \bar{y})^2}\right]\\\\
 &= \text{TSS} - \text{ESS} + 2\,\sum_{j=1}^{n}\,e_j\,(\hat{y}_j - \bar{y})^2\\\\
\end{aligned}
$$

</br>

Such that:

$$
\begin{aligned}
\text{TSS} &\equiv \text{Total Sum of Squares (i.e. deviations from mean)} \\\\
\text{ESS} &\equiv \text{Explained (i.e. by the model) Sum of Squares}\\\\
\text{RSS} &\equiv \text{Residual Sum of Squares}
\end{aligned}
$$

</br>

Both $\text{TSS}$ and $\text{ESS}$ can be directly calculated, however the remaining term $\sum_{j=1}^{n}\,e_j\,(\hat{y}_j - \bar{y})^2$, which is the same in both $r^2(\cdot)$ and $\text{RSS}$, needs further expansion.

$$
\begin{aligned}
\sum_{j=1}^{n}\,e_j\,(\hat{y}_{j} - \bar{y}) &= \sum_{j=1}^{n}\,e_j\,(\beta_0 + \sum_{i=1}^{p}\,\beta_i\,x_{ij} - \bar{y})\\\\
\sum_{j=1}^{n}\,e_j\,(\hat{y}_{j} - \bar{y}) &= \beta_0\,\cancel{\sum_{j=1}^{n}\,e_j}^0 + \sum_{i=1}^{p}\,\beta_i\,\cancel{\sum_{j=1}^{n}\,x_{ij}\,e_j}^0 - \bar{y}\,\cancel{\sum_{j=1}^{n}\,e_j}^0\\\\
\sum_{j=1}^{n}\,e_j\,(\hat{y}_{j} - \bar{y}) &= 0
\end{aligned}
$$

</br>

Therefore, our $r^2(\cdot)$ and $\text{RSS}$ expressions evaluate to:

$$
\begin{aligned}
r^2\,(y, \hat{y}) &= \frac{\left[\text{ESS} + \cancel{\sum_{j=1}^{n}\,e_j\,(\hat{y}_{j} - \bar{y})}^0\right]^2}{\text{TSS}\,\times\,\text{ESS}} \qquad\qquad &\text{RSS} &= \text{TSS} - \text{ESS} + 2\,\cancel{\sum_{j=1}^{n}\,e_j\,(\hat{y}_j - \bar{y})^2}^0 \\\\
r^2\,(y, \hat{y}) &= \frac{\text{ESS}}{\text{TSS}} \qquad\qquad &\text{RSS} &= \text{TSS} - \text{ESS} \\\\
\end{aligned}
$$

</br>

Thus, $r^2(\cdot)$ evaluates either of the two forms:

$$
r^2\,(y, \hat{y}) = \frac{\text{ESS}}{\text{TSS}} \qquad\qquad r^2\,(y, \hat{y}) = 1 - \frac{\text{RSS}}{\text{TSS}}
$$

</br>

*The first form means that the $r^2$ score of a regression model explains how good, in percentage, the model is mimicking the variability of the actual outcomes, as in being compared the best model possible (i.e. one with infinite degrees of freedom so it can perfectly replicate the variations shown in actual data.)*

*The second form shows the $r^2$ score in light of a different perspective: How far a model degrades from being the best one possible (i.e. a $100\%$ model which can perfectly replicate the variations shown in actual data) due to its prediction error variability.*

### 8. On the Meaning of Sample Variance $s^2$

</br>

Sample Variance $s^2$ is different in nature from the Population Variance $\sigma^2$, the first seeks a variance within a sample of size $n$ taken from a population of size $N$. So, let's formulate our expressions:

$$
\begin{aligned}
&\mu_x = \frac{1}{N} \, \sum_{j=1}^{N} \, x_j \qquad\qquad\qquad &\bar{x} &= \frac{1}{n} \, \sum_{j=1}^{n} \, x_j \\\\
&\sigma^2 = \frac{1}{N} \, \sum_{j=1}^{N} \, (x_j - \mu_x)^2
\end{aligned}
$$

</br>

Let's now derive the expression for our sample variance, starting with the Sum of Squares:

$$
\begin{aligned}
\sum_{j=1}^{n} \, (x_j - \bar{x})^2 &= \sum_{j=1}^{n} \, \left[(x_j - \mu_x) - (\bar{x} - \mu_x)\right]^2 \\\\
&= \sum_{j=1}^{n} \, \left[(x_j - \mu_x)^2 - 2 \,(x_j - \mu_x) \,(\bar{x} - \mu_x) + (\bar{x} - \mu_x)^2\right] \\\\
&= \sum_{j=1}^{n} \, (x_j - \mu_x)^2 - 2 \,(\bar{x} - \mu_x)\, \sum_{j=1}^{n} \,(x_j - \mu_x) + \sum_{j=1}^{n} \, (\bar{x} - \mu_x)^2 \\\\
&= \sum_{j=1}^{n} \, (x_j - \mu_x)^2 - 2 \,(\bar{x} - \mu_x)\, (\sum_{j=1}^{n} \,x_j - \sum_{j=1}^{n} \,\mu_x) + n \, (\bar{x} - \mu_x)^2 \\\\
&= \sum_{j=1}^{n} \, (x_j - \mu_x)^2 - 2 \,(\bar{x} - \mu_x)\, (n\,\bar{x} - n\,\mu_x) + n \, (\bar{x} - \mu_x)^2 \\\\
&= \sum_{j=1}^{n} \, (x_j - \mu_x)^2 - 2 \,n\,(\bar{x} - \mu_x)^2 + n \, (\bar{x} - \mu_x)^2 \\\\
&= \sum_{j=1}^{n} \, (x_j - \mu_x)^2 - n \, (\bar{x} - \mu_x)^2
\end{aligned}
$$

</br>

Let's take the Expectation $\mathbb{E}$ for both sides so that we can relax terms that are tricky to evaluate:

$$
\begin{aligned}
\mathbb{E}\left[\sum_{j=1}^{n} \, (x_j - \bar{x})^2\right] &= \mathbb{E}\left[\sum_{j=1}^{n} \, (x_j - \mu_x)^2 - n \, (\bar{x} - \mu_x)^2\right] \\\\
&= \mathbb{E}\left[\sum_{j=1}^{n} \, (x_j - \mu_x)^2 \right] - \mathbb{E}\left[n \, (\bar{x} - \mu_x)^2\right] \\\\
&= \sum_{j=1}^{n} \, \mathbb{E}\left[(x_j - \mu_x)^2\right] - n \, \mathbb{E}\left[(\bar{x} - \mu_x)^2\right]
\end{aligned}
$$

</br>

And since applying the Expectation $\mathbb{E}$ is equivalent to getting a global mean, therefore:

$$
\begin{aligned}
\mathbb{E}\left[x_j\right] &= \mu_x \qquad\qquad \mathbb{E}\left[(x_j - \mu_x)^2\right] = \text{Var}(x) = \sigma^2 \\\\
\mathbb{E}\left[\bar{x}\right] &= \mu_x \quad\longrightarrow\quad \text{A global mean over the means of infinite samples is the population mean}
\end{aligned}
$$

</br>

And since sample means have their own variance from the population mean $\mu_x$:

$$
\begin{aligned}
\mathbb{E}\left[(\bar{x} - \mu_x)^2\right] &= \text{Var}(\bar{x}) \\\\
&= \text{Var}\left(\frac{1}{n}\,\sum_{j=1}^{n}\,x_j\right) \\\\
\end{aligned}
$$

</br>

Let's examine the properties of $\text{Var}$ then. Starting with the property of multiplying the dimension by a constant $c$:

$$
\begin{aligned}
\text{Var}(c\,x) &= \mathbb{E}\left[(c\,x - c\,\mu_x)^2\right] \\\\
&= \mathbb{E}\left[c^2\,(x - \mu_x)^2\right] \\\\
&= c^2\,\mathbb{E}\left[(x - \mu_x)^2\right] \\\\
\end{aligned}
$$

</br>

Therefore:

$$
\begin{aligned}
\text{Var}\left(\frac{1}{n}\,\sum_{j=1}^{n}\,x_j\right) &= \frac{1}{n^2}\,\text{Var}\left(\sum_{j=1}^{n}\,x_j\right)
\end{aligned}
$$

</br>

And now with the summation property:

$$
\begin{aligned}
\text{Var}(x_1 + x_2) &= \mathbb{E}\left[\left((x_1 + x_2) - (\mu_1 + \mu_2)\right)^2\right] \\\\
&= \mathbb{E}\left[\left((x_1 - \mu_1) + (x_2 - \mu_2)\right)^2\right] \\\\
&= \mathbb{E}\left[(x_1 - \mu_1)^2 + 2\,(x_1 - \mu_1)\,(x_2 - \mu_2) + (x_2 - \mu_2)^2\right] \\\\
&= \mathbb{E}\left[(x_1 - \mu_1)^2\right] + 2\,\mathbb{E}\left[(x_1 - \mu_1)\,(x_2 - \mu_2)\right] + \mathbb{E}\left[(x_2 - \mu_2)^2\right] \\\\
&= \text{Var}(x_1) + 2\,\text{Var}(x_1, x_2) + \text{Var}(x_2)
\end{aligned}
$$

</br>

This means that the variance of two feature-dimensions $x_1$ and $x_2$ added together is the sum of their independent variances plus their dependent ones (i.e. covariances). Then again, if $x_1$ and $x_2$ are mutually-orthogonal (i.e. independent observations), the cosine projection of $x_1$ does not exist on $x_2$, and the covariance term $2\,\text{Var}(x_1, x_2)$ vanishes:

$$
\text{Var}(x_1 + x_2) = \text{Var}(x_1) + \text{Var}(x_2)
$$

</br>

Therefore:

$$
\begin{aligned}
\frac{1}{n^2}\,\text{Var}\left(\sum_{j=1}^{n}\,x_j\right) &= \frac{1}{n^2}\,\sum_{j=1}^{n}\,\text{Var}(x_j) \\\\
&= \frac{1}{n^2}\times\,n\,\sigma^2 \\\\
&= \frac{\sigma^2}{n}
\end{aligned}
$$

</br>

Now back to our full list:

$$
\begin{aligned}
\mathbb{E}\left[x_j\right] &= \mu_x \qquad\qquad &\mathbb{E}\left[(x_j - \mu_x)^2\right] &= \sigma^2 \\\\
\mathbb{E}\left[\bar{x}\right] &= \mu_x \qquad\qquad &\mathbb{E}\left[(\bar{x} - \mu_x)^2\right] &= \frac{\sigma^2}{n}
\end{aligned}
$$

</br>

Going back to our original derivation:

$$
\begin{aligned}
\mathbb{E}\left[\sum_{j=1}^{n} \, (x_j - \bar{x})^2\right]  &= \sum_{j=1}^{n} \, \mathbb{E}\left[(x_j - \mu_x)^2\right] - n \, \mathbb{E}\left[(\bar{x} - \mu_x)^2\right] \\\\
&= \sum_{j=1}^{n} \,\text{Var}(x) - n \,\text{Var}(\bar{x}) \\\\
&= n \,\sigma^2 - \cancel{n} \,\frac{\sigma^2}{\cancel{n}} \\\\
&= (n - 1) \,\sigma^2
\end{aligned}
$$

</br>

Therefore:

$$
\begin{aligned}
\sigma^2 &= \mathbb{E}\left[\frac{1}{n-1}\,\sum_{j=1}^{n} \, (x_j - \bar{x})^2\right] \\\\
\sigma^2 &= \mathbb{E}\left[s^2\right]
\end{aligned}
$$

</br>

Where:

$$s^2 = \frac{1}{n-1}\,\sum_{j=1}^{n} \, (x_j - \bar{x})^2$$

*Thus, we have derived the expression for the Sample Variance $s^2$ whilst proving the intuition that taking a global mean of infinite sample variances yields the Population Variance $\sigma^2$*

*An interesting take on the nature of Variance here is that, unlike the population it is taken from, and despite summing through all of them, a sample seems to effectively have less squared deviations to normalize against for representing its variability. So, an $n$-sized sample can be considered as effectively having $n-1$ degrees of freedom, while its $N$-sized population has $N$.*

*Another interesting take is that, like accepting the mean of infinite sample means to become the population mean, we are allowed to accept the hypothesis that the mean of infinite sample variances becomes the population variance if, and only if, two conditions are satisfied: 1) $s^2$ assumes an expression of computation similar to that of $\sigma^2$ (i.e. like $\sigma^2$, $s^2$ must be a normalized sum of squared deviations), and 2) the effective degrees of freedom to normalize against, thereby representing variability in an $n$-sized sample, is to be $n-1$*
